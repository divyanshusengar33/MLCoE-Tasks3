```python
df = df.drop_duplicates()
```

```python
df = df.drop(columns=["SK_ID_CURR"], errors='ignore')
print("Shape after cleaning:", df.shape)
```

```python
missing_percent = df.isnull().mean() * 100
high_missing_columns = missing_percent[missing_percent > 60].index
df = df.drop(columns=high_missing_columns)
```

```python
print("Shape after removing high-missing columns:", df.shape)
```

```python
df.loc[df["AMT_INCOME_TOTAL"] <= 0, "AMT_INCOME_TOTAL"] = np.nan
df.loc[df["AMT_CREDIT"] <= 0, "AMT_CREDIT"] = np.nan
df.loc[df["AMT_ANNUITY"] <= 0, "AMT_ANNUITY"] = np.nan
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())
```

```python
categorical_columns = df.select_dtypes(include="object").columns

for column in categorical_columns:
    df[column] = df[column].fillna(df[column].mode()[0])
```

```python
print("Total missing values:", df.isnull().sum().sum())
```

```python
X = df.drop(columns=["TARGET"])
y = df["TARGET"]
```

```python
print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.preprocessing import OneHotEncoder

categorical_columns = X.select_dtypes(include="object").columns

encoder = OneHotEncoder(
    drop="first",
    sparse_output=False,
    handle_unknown="ignore"
)

encoded = encoder.fit_transform(X[categorical_columns])

encoded_df = pd.DataFrame(
    encoded,
    columns=encoder.get_feature_names_out(categorical_columns),
    index=X.index
)
```

```python
X = X.drop(columns=categorical_columns)
X = pd.concat([X, encoded_df], axis=1)

print(X.shape)
```

```python
from sklearn.model_selection import train_test_split; X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)
print("X_train:", X_train.shape, "X_test:", X_test.shape)
```

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf.fit(X_train, y_train)

importance = pd.Series(
    rf.feature_importances_,
    index=X_train.columns
).sort_values(ascending=False)

top_features = importance.head(30).index

X_train_selected = X_train[top_features]
X_test_selected = X_test[top_features]

print("Selected training features:", X_train_selected.shape)
print("Selected testing features:", X_test_selected.shape)
```

```python
plt.figure(figsize=(10,7))
importance.head(20).sort_values().plot(kind="barh")
plt.xlabel("Importance")
plt.ylabel("Feature")
plt.title("Top 20 Feature Importance")
plt.show()
```
