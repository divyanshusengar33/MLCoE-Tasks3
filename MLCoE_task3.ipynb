```python
df = df.drop_duplicates()
```

```python
df = df.drop(columns=["SK_ID_CURR"], errors='ignore')
print("Shape after cleaning:", df.shape)
```

```python
missing_percent = df.isnull().mean() * 100
high_missing_columns = missing_percent[missing_percent > 60].index
df = df.drop(columns=high_missing_columns)
```

```python
print("Shape after removing high-missing columns:", df.shape)
```

```python
df.loc[df["AMT_INCOME_TOTAL"] <= 0, "AMT_INCOME_TOTAL"] = np.nan
df.loc[df["AMT_CREDIT"] <= 0, "AMT_CREDIT"] = np.nan
df.loc[df["AMT_ANNUITY"] <= 0, "AMT_ANNUITY"] = np.nan
```

```python
numeric_columns = df.select_dtypes(include=np.number).columns

for column in numeric_columns:
    df[column] = df[column].fillna(df[column].median())
```

```python
categorical_columns = df.select_dtypes(include="object").columns

for column in categorical_columns:
    df[column] = df[column].fillna(df[column].mode()[0])
```

```python
print("Total missing values:", df.isnull().sum().sum())
```

```python
X = df.drop(columns=["TARGET"])
y = df["TARGET"]
```

```python
print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
from sklearn.preprocessing import OneHotEncoder

categorical_columns = X.select_dtypes(include="object").columns

encoder = OneHotEncoder(
    drop="first",
    sparse_output=False,
    handle_unknown="ignore"
)

encoded = encoder.fit_transform(X[categorical_columns])

encoded_df = pd.DataFrame(
    encoded,
    columns=encoder.get_feature_names_out(categorical_columns),
    index=X.index
)
```

```python
X = X.drop(columns=categorical_columns)
X = pd.concat([X, encoded_df], axis=1)

print(X.shape)
```

```python
from sklearn.model_selection import train_test_split; X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)
print("X_train:", X_train.shape, "X_test:", X_test.shape)
```

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)

rf.fit(X_train, y_train)

importance = pd.Series(
    rf.feature_importances_,
    index=X_train.columns
).sort_values(ascending=False)

top_features = importance.head(30).index

X_train_selected = X_train[top_features]
X_test_selected = X_test[top_features]

print("Selected training features:", X_train_selected.shape)
print("Selected testing features:", X_test_selected.shape)
```

```python
plt.figure(figsize=(10,7))
importance.head(20).sort_values().plot(kind="barh")
plt.xlabel("Importance")
plt.ylabel("Feature")
plt.title("Top 20 Feature Importance")
plt.show()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train_selected)
X_test_scaled = scaler.transform(X_test_selected)
print(X_train_scaled.shape)
print(X_test_scaled.shape)
```

feature have different ranges and PCA is based on variance this will help us to retain as much as information


```python
from sklearn.decomposition import PCA

pca = PCA()
X_train_pca = pca.fit_transform(X_train_scaled)
X_test_pca = pca.transform(X_test_scaled)
print( X_train_pca.shape)
```

```python
explained_variance = pca.explained_variance_ratio_
print(explained_variance)
```

```python
cumulative_variance = np.cumsum(explained_variance)
print(cumulative_variance)
```

selecting features which expalins 95% of the variance

```python
plt.figure(figsize=(10,5))
plt.plot(range(1, len(cumulative_variance) + 1),
    cumulative_variance,
    marker="o")
plt.axhline(y=0.95, linestyle="--")
plt.xlabel("Number of Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("Cumulative Explained Variance")
plt.show()
```

this graphs tell us about total no. of feature retained after PCA


```python
n_components = np.argmax(cumulative_variance >= 0.95) + 1
print("Number of components:", n_components)
```

```python
pca_final = PCA(n_components=n_components)
X_train_pca = pca_final.fit_transform(X_train_scaled)
X_test_pca = pca_final.transform(X_test_scaled)
```

```python
print("Final PCA training shape:", X_train_pca.shape)
print("Final PCA testing shape:", X_test_pca.shape)
```

PCA is  the way we can represent the data so that we can describe most of its information using fewer features

Our dataset contains many features, and working with a large number of features can make the data more difficult to analyze and visualize. PCA helps reduce these features into a smaller number of components while keeping most of the important information

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(n_estimators=100,random_state=42,n_jobs=-1)
rf_model.fit(X_train_selected, y_train)
rf_pred = rf_model.predict(X_test_selected)
```

Random Forest creates many decision trees using different bootstrap samples of the training data. These trees are trained independently, so they can be built in parallel. For classification, the final prediction is made by taking the majority vote of all the trees. Combining many trees helps reduce overfitting and makes the model more stable

```python
from sklearn.ensemble import AdaBoostClassifier

ada_model = AdaBoostClassifier(n_estimators=100,random_state=42)
ada_model.fit(X_train_selected, y_train)
ada_pred = ada_model.predict(X_test_selected)
```

AdaBoost builds its models one after another instead of building them independently. After each tree, it gives more importance to the samples that were predicted incorrectly. The next tree then focuses more on these difficult samples. In this way, each new tree tries to improve the mistakes made by the previous trees

```python
from xgboost import XGBClassifier

xgb_model = XGBClassifier(n_estimators=100,random_state=42,eval_metric="logloss")
xgb_model.fit(X_train_selected, y_train)
xgb_pred = xgb_model.predict(X_test_selected)
```

XGBoost also builds trees sequentially, but it uses gradient-based optimization to improve the model step by step. It also includes regularization, which helps control model complexity and reduce overfitting. Because of these improvements, XGBoost can often achieve better performance than basic AdaBoost, although the actual performance depends on the dataset and model settings

```python
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)
```

```python
results = []

models = {
    "Random Forest": rf_pred,
    "AdaBoost": ada_pred,
    "XGBoost": xgb_pred
}
```

```python
for name, predictions in models.items():
    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, predictions),
        "Precision": precision_score(y_test, predictions),
        "Recall": recall_score(y_test, predictions),
        "F1 Score": f1_score(y_test, predictions)
    })
```

```python
results_df = pd.DataFrame(results)
```
